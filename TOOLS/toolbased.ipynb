#Ways to create tools
#|
#|
#|---------> using @ tool decorator
#|
#|---------> using StructuredTool & Pydantic
#|
#|---------> using BaseTool class


```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class AdditionInput(BaseModel):
    a : int = Field(required=True, description="The first number to be add")
    b : int = Field(required=True, description="The second number to be add")
```

```python
def add_func(a: int, b: int) -> int:
    return a + b
```

```python
add_tool = StructuredTool.from_function(
    func=add_func,
    name='Add Karo',
    description='Add karte raho !!',
    args_schema=AdditionInput
)
```

```python
result = add_tool.invoke({'a':68, 'b':1})
print(result)
```
